# Jupyter & Scientific Workflow
This notebook introduces basic Jupyter usage, inline plotting and simple data handling.

## Step 1: Import libraries.
We'll use Pandas for data handling and matplotlib for plotting.
Note that matplotlib will be inlined so as to embed plots in the current notebook.

````python
import pandas as pd             # Data handling
import numpy as np              # Data generation
import matplotlib.pyplot as plt # Plotting
````

## Step 2: Create sample dataset.
A small DataFrame object with some x and y values will be created. A function will be cerated for such purpose.

````python
def generateRandomData( min_x:          int     = 0     ,
                        max_x:          int     = 10    ,
                        num_of_points:  int     = 100   ,
                        mean_value:     float   = 0.0   ,
                        std_deviation:  float   = 0.2   ,
                        random_seed:    int     = 33    ) -> pd.DataFrame:
    # Use a seed so as to keep results reproducibility.
    np.random.seed(random_seed)
    
    # Generate data.
    x: np.ndarray = np.linspace(min_x, max_x, num_of_points)
    y: np.ndarray = np.sin(x) + np.random.normal(loc = mean_value, scale = std_deviation, size = len(x))

    # Create DataFrame out of the data generated above.
    df: pd.DataFrame = pd.DataFrame({"x" : x, "y" : y})
    df.set_index(x, inplace = True)

    return df
````

## Step 3: Plot data.
Use Matplotlib library to plot data in a humanly-friendly format.

````python
def plotData(   df              :   pd.DataFrame            ,
                x_axis_data_name:   str = "x"               ,
                x_axis_label    :   str = "X"               ,
                y_axis_data_name:   str = "y"               ,
                y_axis_label    :   str = "Y"               ,
                plot_title      :   str = "Random data plot",
                data_marker     :   str = 'o'               ) -> None:
    plt.plot(df[x_axis_data_name], df[y_axis_data_name], marker = data_marker)
    plt.title(plot_title)
    plt.xlabel(x_axis_label)
    plt.ylabel(y_axis_label)
    plt.grid(True)
    plt.show()
````

## Step 4: Execute data generator and data plotting functions.
Just add some code structure.

````python
def main():
    plotData(generateRandomData(), x_axis_label = "t (s)", y_axis_label = "Noisy sin(t) signal")

if __name__ == "__main__":
    main()
````

### Noisy Sine Wave
The data generated is modeled as:

$$
y = \sin(x) + \epsilon, \quad \epsilon \sim \mathcal{N}(0, \sigma^2)
$$
