# Interactive widgets
Jupyter Notebooks allow interactive widgets to be embedded within. To do so, _@interactive_ decorator can be used. Note that such _@interactive_ decorator is solely understandable for the interpreter when called in a notebook's context.

We will start by importing required modules (ipywidgets so as to acquire _@interactive_ decorator, NumPy and the 1D series plotting function) as well as adding _widget_ magic word before everything else.

```python
# Choose interactive backend
%matplotlib widget
from ipywidgets import interact, FloatSlider, Checkbox, ColorPicker, Dropdown
import numpy as np
from utils import plot1DTimeSeries
import matplotlib.pyplot as plt
```

## Add an interactive-decorated function
In the case of the function below, a simple slider control will be added to the generated widget-like plot. It will alow to choose a sine-wave's amplitude. So as to generate the slider control, _@interact_ decorator should be called before the function by providing min, max values and the step between slider's selectable values.

Additionally, some other controllers will be added (line color selector, grid on/off button, waveform selector...).

```python
@interact(amplitude = FloatSlider(min = 0.5, max = 3.0, step = 0.1, value = 1.0, description = "Amplitude"),
          frequency = FloatSlider(min = 0.5, max = 3.0, step = 0.1, value = 1.0, description = "Frequency"),
          waveform  = Dropdown(options = ["sin", "cos"], value = "sin", description = "Waveform"),
          show_grid = Checkbox(value=True, description="Show Grid"),
          line_color= ColorPicker(value = "Blue", description = "Line Color"))
def plot_signal(amplitude, frequency, waveform, show_grid, line_color):
    t       : np.ndarray = np.linspace(0, 10, 100)
    signal  : np.ndarray = amplitude * (np.sin(t * frequency) if waveform == "sin" else np.cos(t * frequency))

    plot1DTimeSeries(signal, time_step = 0.1, title = "Interactive Sine Wave")
    
    # Change the line color on the fly
    plt.gca().lines[-1].set_color(line_color)
    plt.grid(show_grid)
    plt.draw()
```
